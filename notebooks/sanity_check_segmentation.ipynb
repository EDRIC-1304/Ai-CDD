```python
import os
import random
import cv2
import matplotlib.pyplot as plt

```

```python
IMAGE_DIR = r"D:\FY_Project\Ai-CDD\data\raw\segmentation_sample\images"
MASK_DIR  = r"D:\FY_Project\Ai-CDD\data\raw\segmentation_sample\masks"

print("Image dir exists:", os.path.exists(IMAGE_DIR))
print("Mask dir exists:", os.path.exists(MASK_DIR))
```

```python
image_files = sorted(os.listdir(IMAGE_DIR))
print("Total images found:", len(image_files))

sample_files = random.sample(image_files, 5)
print("Samples selected:", sample_files)
```

```python
for img_name in sample_files:
    img_path = os.path.join(IMAGE_DIR, img_name)
    mask_path = os.path.join(
        MASK_DIR, img_name.replace(".png", "_mask.png")
    )

    image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(8, 4))

    plt.subplot(1, 2, 1)
    plt.title("X-ray Image")
    plt.imshow(image, cmap="gray")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.title("Lung Mask")
    plt.imshow(mask, cmap="gray")
    plt.axis("off")

    plt.show()
```
